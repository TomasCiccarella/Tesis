```python
import numpy as np
import matplotlib.pyplot as plt

def potencial(phi, V0, alpha):
    return V0 * (1 - np.exp(- np.abs(phi) / (np.sqrt(6) * alpha)))**2

alpha = 1
V0 = 1e-10
phi = np.linspace(-20, 3, 4000)
V = potencial(phi, V0, alpha)

plt.figure(figsize=(25, 10))
plt.plot(phi, V, lw=10)
plt.scatter(-3, potencial(-3, V0, alpha), color='red', zorder=20, s=500)
plt.scatter(-18, potencial(-18, V0, alpha) + 3e-12, color='red', zorder=20, s=500)
plt.axvline(x=-3, ymin=0, ymax=0.5, color='black', lw=5, ls="--", alpha=0.7)
plt.text(-3.5, -1.7e-11, r'$\phi_{end}$', fontsize=75)
plt.annotate('', xy=(-17, potencial(-18, V0, alpha) + 3e-12), xytext=(-18, potencial(-18, V0, alpha) + 3e-12),
             arrowprops=dict(arrowstyle="->", color='red', lw=2))
plt.grid(alpha=0.3)
plt.xticks([])
plt.yticks([])
plt.xlabel(r'$\phi$', fontsize=80)
plt.ylabel(r'$V(\phi)$', fontsize=80, rotation=0, labelpad=90)

# Agregar las llaves con anotaciones para las zonas

# Zona Inflación
plt.text(-10.5, 1.1e-10, "Inflación", color="green", fontsize=75, ha="center")
plt.axvspan(-20, -3, color='green', alpha=0.3)

# Zona Reheating
plt.text(0, 1.1e-10, "Reheating", color="DarkOrange", fontsize=75, ha="center")
plt.axvspan(-3, 3, color='orange', alpha=0.3)

# plt.show()
plt.savefig("reheating", format='pdf')
```
