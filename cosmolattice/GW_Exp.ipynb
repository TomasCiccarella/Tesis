```python
H0 = 67.4 * 1e3 / (3.086e22)  # Hubble constant in SI units (s^-1)

# LC Circuits
xEMRC = np.concatenate((x1, x2))
yEMRC = np.concatenate((y1, y2))
xEMRC = 10**xEMRC
yEMRC = 10**yEMRC
OmegaEMRC = (2 * np.pi * xEMRC * yEMRC/(np.sqrt(3) * H0))**2
hc = np.sqrt(3 * H0**2 * OmegaEMRC / (4 * np.pi**2 * xEMRC**2))

plt.figure(figsize=(12, 6))
plt.plot(xEMRC, OmegaEMRC, 'k-', label='EMRC', linewidth=2)
#plt.plot(xEMRC, hc, 'r-', label='Characteristic Strain', linewidth=2)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.xlabel(r'$f_0$ [Hz]', fontsize=14)
plt.ylabel(r'$\Omega_{GW}$', fontsize=14)
plt.xscale('log')
plt.yscale('log')
plt.tight_layout()
```

```python
H0 = 67.4 * 1e3 / (3.086e22)  # Hubble constant in SI units (s^-1)

# LC Circuits
xLC1 = np.logspace(3.5,4.8,1000)
xLC2 = np.logspace(4.8,5.2,1000)
xLC3 = np.logspace(5.2,8.5,1000)
yLC1 = np.logspace(-10,-12,1000)
yLC2 = np.logspace(-12,-16,1000)
yLC3 = np.logspace(-16,-20,1000)
xLC = np.concatenate((xLC1, xLC2, xLC3))
yLC = np.concatenate((yLC1, yLC2, yLC3))
OmegaLC = (2 * np.pi * xLC * yLC/(np.sqrt(3) * H0))**2
hc = np.sqrt(3 * H0**2 * OmegaLC / (4 * np.pi**2 * xLC**2))

plt.figure(figsize=(12, 8))
plt.plot(xLC, OmegaLC, 'k-', label='LC Circuits', linewidth=2)
plt.plot(xLC, hc, 'r-', label='Characteristic Strain', linewidth=2)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.xlabel(r'$f_0$ [Hz]', fontsize=14)
plt.ylabel(r'$\Omega_{GW}$', fontsize=14)
plt.xscale('log')
plt.yscale('log')
plt.tight_layout()
```
