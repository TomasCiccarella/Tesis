```python
dir_m2phi2 = "./build_m2phi2/strain_spectrum_m2phi2.txt"
dir_lphi4 = "./build_lphi4/strain_spectrum_lphi4.txt"
dir_tanh2 = "./build_tanh2/strain_spectrum_tanh2.txt"
dir_tanh4 = "./build_tanh4/strain_spectrum_tanh4.txt"
dir_Starobinsky = "./build_starobinsky3/strain_spectrum_Starobinsky.txt"

data_m2phi2 = np.loadtxt(dir_m2phi2)
data_lphi4 = np.loadtxt(dir_lphi4)
data_tanh2 = np.loadtxt(dir_tanh2)
data_tanh4 = np.loadtxt(dir_tanh4)
data_Starobinsky = np.loadtxt(dir_Starobinsky)

plt.figure(figsize=(12, 8))
plt.plot(data_m2phi2[:, 0], data_m2phi2[:, 1], 'r-', label=r'$m^2\phi^2$')
plt.loglog(data_lphi4[:, 0], data_lphi4[:, 1], 'b-', label=r'$\lambda\phi^4$')
plt.loglog(data_tanh2[:, 0], data_tanh2[:, 1], 'g-', label=r'$\tanh^2$')
plt.loglog(data_tanh4[:, 0], data_tanh4[:, 1], 'm-', label=r'$\tanh^4$')
plt.loglog(data_Starobinsky[:, 0], data_Starobinsky[:, 1], 'c-', label=r'$\text{Starobinsky}$')
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.xlabel(r'Frequency $f_0$ [Hz]', fontsize=14)
plt.ylabel(r'Characteristic Strain $h_c(f_0, t_0)$', fontsize=14)
plt.title('Gravitational Wave Strain Spectrum', fontsize=16)
plt.legend()
plt.tight_layout()
plt.show()
```
