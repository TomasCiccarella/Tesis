# Comparación gráfica de CI's

```python
import numpy as np
import matplotlib.pyplot as plt
```

La posición inicial la calcularemos a partir del parámetro $\epsilon$ que se define como:

$$\epsilon = \frac{m_p^2}{2} \left[ \frac{\partial_\phi V(\phi)}{V(\phi)} \right]^2$$

y pediremos que en el momento inicial $\epsilon (\phi_0) = 1$.

A su vez, la condición inicial de la velocidad la podemos obtener a partir de pedir que la energía cinética del campo sea comparable con la energía potencial lo que implica que tenemos que pedir la siguiente condición:

$$\dot{\phi}_0 = - \sqrt{2 V(\phi_0)}$$

## $m^2 \phi^2$ vs $\tanh^2$ vs Starobinsky

Para $m^2 \phi^2$ el potencial es de la forma $\displaystyle \frac{1}{2} m^2 \phi^2$ por tanto la condición que nos dá la posición inicial es:

$$\begin{split}
    \frac{m_p^2}{2} \left( \frac{m^2 \phi_0}{\frac{1}{2} m^2 \phi_0^2} \right)^2 &= 1\\
    \frac{m_p^2}{2} \left( \frac{2}{\phi_0} \right)^2 &= 1\\
    \frac{2 m_p^2}{\phi_0^2} &= 1
\end{split}$$

$$\phi_0 = \sqrt{2} m_p$$

Y la condición para la velocidad inicial sería:

$$\begin{split}
    \dot{\phi}_0 &= - \sqrt{m^2 \phi_0^2}\\
    \dot{\phi}_0 &= - m \phi_0
\end{split}$$

$$\dot{\phi}_0 = - \sqrt{2} m m_p $$

Para $\tanh^2$ tenemos que el potencial es de la forma $\displaystyle \frac{1}{2} m^2 \mu^2 \tanh^2 \left( \frac{\phi}{\mu} \right)$ por tanto la condición inicial del campo va a estar dada por:

$$\begin{split}
    \frac{m_p^2}{2} \left[ \frac{m^2 \mu \tanh \left( \frac{\phi_0}{\mu} \right) \text{sech}^2 \left( \frac{\phi_0}{\mu} \right)}{\frac{1}{2} m^2 \mu^2 \tanh^2 \left( \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{m_p^2}{2} \left[ \frac{2 \text{sech}^2 \left( \frac{\phi_0}{\mu} \right)}{\mu \tanh \left( \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{m_p^2}{2} \left[ \frac{2}{\mu \sinh \left( \frac{\phi_0}{\mu} \right) \cosh \left( \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{m_p^2}{2} \left[ \frac{4}{\mu \sinh \left( 2 \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{8 m_p^2}{\mu^2 \sinh^2 \left( 2 \frac{\phi_0}{\mu} \right)} &= 1\\
    \sinh^2 \left( 2 \frac{\phi_0}{\mu} \right) &= 8 \left( \frac{m_p}{\mu} \right)^2\\
    2 \frac{\phi_0}{\mu} &= \text{arcsinh} \left( 2 \sqrt{2} \frac{m_p}{\mu} \right)
\end{split}$$

$$\phi_0 = \frac{\mu}{2} \text{arcsinh} \left( 2 \sqrt{2} \frac{m_p}{\mu} \right)$$

Y la velocidad inicial del campo nos quedará:

$$\begin{split}
    \dot{\phi}_0 &= -\sqrt{m^2 \mu^2 \tanh^2 \left( \frac{\phi_0}{\mu} \right)}\\
    \dot{\phi}_0 &= - m \mu \tanh \left( \frac{\phi_0}{\mu} \right)
\end{split}$$

$$\dot{\phi}_0 = - m \mu \tanh \left[ \frac{1}{2} \text{arcsinh} \left( 2 \sqrt{2} \frac{m_p}{\mu} \right) \right]$$

Para Starobinsky, el potencial es de la forma $\displaystyle \frac{1}{2} m^2 \mu^2 \left[ 1 - \exp \left(- \frac{\phi}{\mu} \right) \right]^2$ por tanto la condición inicial del campo vale

$$\begin{split}
    \frac{m_p^2}{2} \left\{ \frac{m^2 \mu \left[ 1 - \exp \left( - \frac{\phi_0}{\mu} \right) \right] \exp \left( - \frac{\phi_0}{\mu} \right)}{\frac{1}{2} m^2 \mu^2 \left[ 1 - \exp \left(- \frac{\phi_0}{\mu} \right) \right]^2} \right\}^2 &= 1\\
    \frac{m_p^2}{2} \left\{ \frac{2 \exp \left(- \frac{\phi_0}{\mu} \right)}{\mu \left[ 1 - \exp \left(- \frac{\phi_0}{\mu} \right) \right]} \right\}^2 &= 1\\
    \frac{m_p^2}{2} \left\{ \frac{2}{\mu \left[ \exp \left(\frac{\phi_0}{\mu} \right) - 1 \right]} \right\}^2 &= 1\\
    \frac{2 m_p^2}{\mu^2 \left[ \exp \left(\frac{\phi_0}{\mu} \right) - 1 \right]^2} &= 1\\ 
    \left[ \exp \left(\frac{\phi_0}{\mu} \right) - 1 \right]^2 &= 2 \left( \frac{m_p}{\mu} \right)^2\\
    \exp \left( \frac{\phi_0}{\mu} \right) &= \sqrt{2} \frac{m_p}{\mu} + 1\\
    \frac{\phi_0}{\mu} &= \ln \left( \sqrt{2} \frac{m_p}{\mu} + 1 \right) 
\end{split}$$

$$\phi_0 = \mu \ln \left( \sqrt{2} \frac{m_p}{\mu} + 1 \right)$$

Y la velocidad inicial del campo valdrá

$$\begin{split}
    \dot{\phi}_0 &= -\sqrt{m^2 \mu^2 \left[ 1 - \exp \left(- \frac{\phi_0}{\mu} \right) \right]^2}\\
    \dot{\phi}_0 &= -m \mu \left[ 1 - \exp \left( - \frac{\phi_0}{\mu} \right) \right]\\
    \dot{\phi}_0 &= -m \mu \left\{ 1 - \exp \left[ - \ln \left( \sqrt{2} \frac{m_p}{\mu} + 1 \right) \right] \right\}\\
    \dot{\phi}_0 &= -m \mu \left( 1 - \frac{1}{\sqrt{2} \frac{m_p}{\mu} + 1} \right)\\
    \dot{\phi}_0 &= -m \mu \frac{\sqrt{2} \frac{m_p}{\mu}}{\sqrt{2} \frac{m_p}{\mu} + 1}\\
\end{split}$$

$$\dot{\phi}_0 = - \frac{\sqrt{2} m \mu m_p }{\sqrt{2} m_p + \mu}$$

```python
def m2phi2(m, phi):
    return 0.5 * m**2 * phi**2

def tanh2(m, mu, phi):
    return 0.5 * m**2 * mu**2 * np.tanh(phi/mu)**2

def starobinsky3(m, mu, phi):
    return 0.5 * m**2 * mu**2 * (1 - np.exp(-phi/mu))**2

def dm2phi2_dphi(m, phi):
    return m**2 * phi

def dtanh2_dphi(m, mu, phi):
    return m**2 * mu * np.tanh(phi/mu) * (1 / np.cosh(phi/mu))**2

def dstarobinsky3_dphi(m, mu, phi):
    return m**2 * mu * (1 - np.exp(-phi/mu)) * np.exp(-phi/mu)
```

```python
m = 1 #2.435e19
mp = 1 #2.435e19
lambda4 = 1 #1.7966e65
mu = np.sqrt(2/3) * mp #np.sqrt(lambda4)/m

phi = np.linspace(-2*mp, 3*mp, 100000)

plt.figure(figsize=(20, 8))
plt.plot(phi, m2phi2(m, phi), label=r'$m^2 \phi^2$')
plt.plot(phi, tanh2(m, mu, phi), label=r'$\tanh^2$')
plt.plot(phi, starobinsky3(m, mu, phi), label=r'Starobinsky3')
plt.xlabel(r'$\tilde{\phi}$')
plt.ylabel(r'$\tilde{V}(\phi)$')
plt.ylim(0, 1)
plt.legend()
plt.title('Comparación de potenciales')
plt.tight_layout()
plt.show();
```

```python
phi0_m2phi2 = np.sqrt(2)*mp
phi0_tanh2 = mu/2 * np.arcsinh(2*np.sqrt(2)*mp/mu)
phi0_starobinsky3 = mu*np.log(1 + np.sqrt(2)*mp/mu)

print(f"phi0 m2phi2: {phi0_m2phi2}")
print(f"phi0 tanh2: {phi0_tanh2}")
print(f"phi0 starobinsky3: {phi0_starobinsky3}")

dphi0_m2phi2 = -1 * np.sqrt(m**2 * phi0_m2phi2**2)
dphi0_tanh2 = -1 * np.sqrt(m**2 * mu**2 * np.tanh(phi0_tanh2/mu)**2)
dphi0_starobinsky3 = -1 * np.sqrt(m**2 * mu**2 * (1 - np.exp(-phi0_starobinsky3/mu))**2)

print(f"dphi0 m2phi2: {dphi0_m2phi2}")
print(f"dphi0 tanh2: {dphi0_tanh2}")
print(f"dphi0 starobinsky3: {dphi0_starobinsky3}")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(phi, m2phi2(m, phi), label=r'$m^2 \phi^2$', lw = 5, color = 'blue')
plt.plot(phi, tanh2(m, mu, phi), label=r'$\tanh^2$', lw = 5, color = 'green')
plt.plot(phi, starobinsky3(m, mu, phi), label=r'Starobinsky', lw = 5, color = 'red')
plt.scatter(phi0_m2phi2, m2phi2(m, phi0_m2phi2), color='blue', s = 200)
plt.scatter(phi0_tanh2, tanh2(m, mu, phi0_tanh2), color='green', s = 200)
plt.scatter(phi0_starobinsky3, starobinsky3(m, mu, phi0_starobinsky3), color='red', s = 200)

# Flechas tangentes con módulo proporcional a |dphi0|
scale = 0.1  # Factor de escala para la longitud de las flechas

# Para m2phi2
slope = dm2phi2_dphi(m, phi0_m2phi2)
vec_x = -1
vec_y = -slope
norm = np.sqrt(vec_x**2 + vec_y**2)
dx = (vec_x / norm) * abs(dphi0_m2phi2) * scale
dy = (vec_y / norm) * abs(dphi0_m2phi2) * scale
plt.arrow(phi0_m2phi2, m2phi2(m, phi0_m2phi2), dx, dy, head_width=0.02, head_length=0.02, fc='black', ec='black', zorder = 20, lw = 3)

# Para tanh2
slope = dtanh2_dphi(m, mu, phi0_tanh2)
vec_x = -1
vec_y = -slope
norm = np.sqrt(vec_x**2 + vec_y**2)
dx = (vec_x / norm) * abs(dphi0_tanh2) * scale
dy = (vec_y / norm) * abs(dphi0_tanh2) * scale
plt.arrow(phi0_tanh2, tanh2(m, mu, phi0_tanh2), dx, dy, head_width=0.02, head_length=0.02, fc='black', ec='black', zorder = 20, lw = 3)

# Para starobinsky3
slope = dstarobinsky3_dphi(m, mu, phi0_starobinsky3)
vec_x = -1
vec_y = -slope
norm = np.sqrt(vec_x**2 + vec_y**2)
dx = (vec_x / norm) * abs(dphi0_starobinsky3) * scale
dy = (vec_y / norm) * abs(dphi0_starobinsky3) * scale
plt.arrow(phi0_starobinsky3, starobinsky3(m, mu, phi0_starobinsky3), dx, dy, head_width=0.02, head_length=0.02, fc='black', ec='black', zorder = 20, lw = 3)

plt.xlabel(r'$\tilde{\phi}$', fontsize = 20)
plt.ylabel(r'$\tilde{V}(\tilde\phi)$', fontsize = 20, labelpad = 20, rotation = 0)
plt.ylim(0, 1.25)
plt.legend(fontsize = 16)
plt.xticks(fontsize = 16)
plt.yticks(fontsize = 16)
#plt.title('Comparación de potenciales', fontsize = 26)
plt.tight_layout()
#plt.show();
plt.savefig("comparacion_w0.pdf", format='pdf', bbox_inches='tight')
```

## $\lambda \phi^4$ vs $\tanh^4$

Para $\lambda \phi^4$ el potencial es de la forma $\displaystyle \frac{1}{4} \lambda \phi^4$ por tanto la condición que nos dá la posición inicial es:

$$\begin{split}
    \frac{m_p^2}{2} \left( \frac{\lambda \phi_0^3}{\frac{1}{4} \lambda \phi_0^4} \right)^2 &= 1\\
    \frac{m_p^2}{2} \left( \frac{4}{\phi_0} \right)^2 &= 1\\
    \frac{8 m_p^2}{\phi_0^2} &= 1
\end{split}$$

$$\phi_0 = 2 \sqrt{2} m_p$$

Y la condición para la velocidad inicial sería:

$$\begin{split}
    \dot{\phi}_0 &= - \sqrt{\frac{1}{2} \lambda \phi_0^4}\\
    \dot{\phi}_0 &= - \sqrt{32 \lambda m_p^4}
\end{split}$$

$$\dot{\phi}_0 = - 4 \sqrt{2 \lambda} m_p^2$$

Para $\tanh^4$ tenemos que el potencial es de la forma $\displaystyle \frac{1}{4} m^2 \mu^2 \tanh^4 \left( \frac{\phi}{\mu} \right)$ por tanto la condición inicial del campo va a estar dada por:

$$\begin{split}
    \frac{m_p^2}{2} \left[ \frac{m^2 \mu \tanh^3 \left( \frac{\phi_0}{\mu} \right) \text{sech}^2 \left( \frac{\phi_0}{\mu} \right)}{\frac{1}{4} m^2 \mu^2 \tanh^4 \left( \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{m_p^2}{2} \left[ \frac{4 \text{sech}^2 \left( \frac{\phi_0}{\mu} \right)}{\mu \tanh \left( \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{m_p^2}{2} \left[ \frac{4}{\mu \sinh \left( \frac{\phi_0}{\mu} \right) \cosh \left( \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{m_p^2}{2} \left[ \frac{8}{\mu \sinh \left( 2 \frac{\phi_0}{\mu} \right)} \right]^2 &= 1\\
    \frac{32 m_p^2}{\mu^2 \sinh^2 \left( 2 \frac{\phi_0}{\mu} \right)} &= 1\\
    \sinh^2 \left( 2 \frac{\phi_0}{\mu} \right) &= 32 \left( \frac{m_p}{\mu} \right)^2\\
    2 \frac{\phi_0}{\mu} &= \text{arcsinh} \left( 4 \sqrt{2} \frac{m_p}{\mu} \right)
\end{split}$$

$$\phi_0 = \frac{\mu}{2} \text{arcsinh} \left( 4 \sqrt{2} \frac{m_p}{\mu} \right)$$

Y la velocidad inicial del campo nos quedará:

$$\begin{split}
    \dot{\phi}_0 &= -\sqrt{\frac{1}{2} m^2 \mu^2 \tanh^4 \left( \frac{\phi_0}{\mu} \right)}\\
    \dot{\phi}_0 &= - \sqrt{\frac{1}{2}} m \mu \tanh^2 \left( \frac{\phi_0}{\mu} \right)
\end{split}$$

$$\dot{\phi}_0 = - \sqrt{\frac{1}{2}} m \mu \tanh^2 \left[ \frac{1}{2} \text{arcsinh} \left( 4 \sqrt{2} \frac{m_p}{\mu} \right) \right]$$

```python
def lphi4(l, phi):
    return 0.25 * l * phi**4

def tanh4(m, mu, phi):
    return 0.25 * m**2 * mu**2 * np.tanh(phi/mu)**4

def dlphi4_dphi(l, phi):
    return l * phi**3

def dtanh4_dphi(m, mu, phi):
    return m**2 * mu * np.tanh(phi/mu)**3 * (1 / np.cosh(phi/mu))**2
```

```python
m = 1 #2.435e19
mp = 1 #2.435e19
l = 1 #1.7966e65
mu = np.sqrt(2/3) * mp #np.sqrt(lambda4)/m

phi = np.linspace(-2*mp, 3*mp, 100000)

plt.figure(figsize=(20, 8))
plt.plot(phi, lphi4(l, phi), label=r'$\lambda \phi^4$')
plt.plot(phi, tanh4(m, mu, phi), label=r'$\tanh^4$')
plt.xlabel(r'$\tilde{\phi}$')
plt.ylabel(r'$\tilde{V}(\phi)$')
plt.ylim(0, 1)
plt.legend()
plt.title('Comparación de potenciales')
plt.tight_layout()
plt.show();
```

```python
phi0_lphi4 = 2*np.sqrt(2)*mp
phi0_tanh4 = mu/2 * np.arcsinh(4*np.sqrt(2)*mp/mu)

print(f"phi0 lphi4: {phi0_lphi4}")
print(f"phi0 tanh4: {phi0_tanh4}")

dphi0_lphi4 = -1 * np.sqrt(0.5 * l * phi0_lphi4**4)
dphi0_tanh4 = -1 * np.sqrt(0.5 * m**2 * mu**2 * np.tanh(phi0_tanh4/mu)**4)

print(f"dphi0 lphi4: {dphi0_lphi4}")
print(f"dphi0 tanh4: {dphi0_tanh4}")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(phi, lphi4(l, phi), label=r'$\lambda \phi^4$', lw = 5, color = 'orange')
plt.plot(phi, tanh4(m, mu, phi), label=r'$\tanh^4$', lw = 5, color = 'purple')
plt.scatter(phi0_lphi4, lphi4(l, phi0_lphi4), color='orange', s = 200)
plt.scatter(phi0_tanh4, tanh4(m, mu, phi0_tanh4), color='purple', s = 200)

# Flechas tangentes con módulo proporcional a |dphi0|
scale = 0.1  # Factor de escala para la longitud de las flechas

# Para lphi4
slope = dlphi4_dphi(m, phi0_lphi4)
vec_x = -1
vec_y = -slope
norm = np.sqrt(vec_x**2 + vec_y**2)
dx = (vec_x / norm) * abs(dphi0_lphi4) * scale
dy = (vec_y / norm) * abs(dphi0_lphi4) * scale
plt.arrow(phi0_lphi4, lphi4(l, phi0_lphi4), dx, dy, head_width=0.02, head_length=0.02, fc='black', ec='black', zorder = 20, lw = 3)

# Para tanh4
slope = dtanh4_dphi(m, mu, phi0_tanh4)
vec_x = -1
vec_y = -slope
norm = np.sqrt(vec_x**2 + vec_y**2)
dx = (vec_x / norm) * abs(dphi0_tanh4) * scale
dy = (vec_y / norm) * abs(dphi0_tanh4) * scale
plt.arrow(phi0_tanh4, tanh4(m, mu, phi0_tanh4), dx, dy, head_width=0.02, head_length=0.02, fc='black', ec='black', zorder = 20, lw = 3)

plt.xlabel(r'$\tilde{\phi}$', fontsize = 20)
plt.ylabel(r'$\tilde{V}(\tilde\phi)$', fontsize = 20, labelpad = 20, rotation = 0)
plt.yscale("log")
plt.legend(fontsize = 16)
plt.xticks(fontsize = 16)
plt.yticks(fontsize = 16)
#plt.title('Comparación de potenciales', fontsize = 26)
plt.tight_layout()
#plt.show();
plt.savefig("comparacion_w13.pdf", format='pdf', bbox_inches='tight')
```
